```python
from ngsolve import *
from netgen.geom2d import unit_square
mesh = Mesh(unit_square.GenerateMesh(maxh=0.1))
fes = H1(mesh, order=1)
u,v = fes.TnT()
a = BilinearForm(grad(u)*grad(v)*dx+10*u*v*dx).Assemble()
f = LinearForm(x*y*v*dx).Assemble()
gfu = GridFunction(fes)

r = f.vec.CreateVector()
p_old = f.vec.CreateVector()
p_new = f.vec.CreateVector()
p1 = f.vec.CreateVector()

gfu.vec[:] = 0
r.data = f.vec
p_old.data = r.data
err0 = Norm(r)
its = 0

while True:
    p1.data = a.mat * p_old
    err2 = InnerProduct(r,r)
    alpha = err2 / InnerProduct(p_old,p1)

    print ("iteration", its, "res=", sqrt(err2))
    gfu.vec.data += alpha * p_old
    r.data -= alpha * p1
    
    err3 = InnerProduct(r,r)
    beta = - err3 / err2
    p_new.data = r - beta * p_old.data
    p_old.data = p_new.data
    if sqrt(err3) < 1e-8 * err0 or its > 1000: break
    its = its+1
print ("needed", its, "iterations")
```
